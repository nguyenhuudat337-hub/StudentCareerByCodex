## 1. Nạp CSV và kiểm tra cấu trúc, các kiểu dữ liệu

## 2. Kiểm tra missing và duplicates (bắt buộc dù không có lỗi)

Không xóa hoặc tự điền dữ liệu mà chưa có lý do.

```python
print("Missing từng cột:")
print(df.isnull().sum())
print("Tổng missing:",df.isnull().sum().sum())
print("Duplicates toàn dòng:",df.duplicated().sum())
print("Duplicates bỏ ID:",df.drop(columns="student_id",errors="ignore").duplicated().sum())
print("Số lớp nhãn:",df["recommended_career"].nunique())
display(df["recommended_career"].value_counts())
audit = audit_data(df)
print("Các biến số có Inf:", {k:v["infinite"] for k,v in audit["numeric_ranges"].items() if v["infinite"]})
```
